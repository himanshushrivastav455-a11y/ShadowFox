## BERT Language Model for Sentiment Analysis

```
from transformers import pipeline
import pandas as pd
import matplotlib.pyplot as plt

classifier = pipeline("sentiment-analysis",
                      model="distilbert-base-uncased-finetuned-sst-2-english")
print("Model Loaded Successfully")
```

```
texts=[
"I love Artificial Intelligence.",
"This internship is amazing.",
"I hate waiting in traffic.",
"The movie was boring.",
"ChatGPT is very helpful.",
"The food was terrible.",
"The weather is wonderful today.",
"I am disappointed with the service."
]

results=[]
for t in texts:
    p=classifier(t)[0]
    results.append({"Text":t,"Sentiment":p["label"],"Score":p["score"]})
df=pd.DataFrame(results)
df
```

```
counts=df["Sentiment"].value_counts()
counts.plot(kind="bar",title="Sentiment Distribution")
plt.show()

counts.plot(kind="pie",autopct="%1.1f%%",ylabel="")
plt.title("Sentiment Distribution")
plt.show()
```
